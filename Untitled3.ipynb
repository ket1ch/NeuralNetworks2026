испортируем нужные библиотеки для трансфер лернинга

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader, random_split
import os
import matplotlib.pyplot as plt
from PIL import ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Используется устройство: {device}")
```

смотрим какие классы, типы грибов есть

```python
DATA_DIR = "/kaggle/input/mushrooms-classification-common-genuss-images/Mushrooms"
print("Классы:", os.listdir(DATA_DIR))
```

трансформации с аугументациями и без них

```python
train_transforms = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
val_transforms = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
full_dataset = datasets.ImageFolder(root=DATA_DIR)
num_classes = len(full_dataset.classes)
print(f"Найдено классов: {num_classes}")
print(f"Названия классов: {full_dataset.classes}")
print(f"Всего изображений: {len(full_dataset)}")
```

делим данные на тренировочные и валидационные (с соотнощениями 80 к 20)

```python
train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])
```

```python
train_dataset.dataset.transform = train_transforms
val_dataset.dataset.transform = val_transforms
```

создаем даталоадеры

```python
batch_size = 32
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=2)
```

создаем модель трансфер лернинга
тут мы заморозим слои так как датасет не очень большой и модель просто может заучить картинки

```python
model = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
for param in model.parameters():
    param.requires_grad = False
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, num_classes)
model = model.to(device)
```

начинаем обучение

```python
criterion = nn.CrossEntropyLoss()
# Обучаем только последний слой (так как остальное заморожено)
optimizer = optim.Adam(model.fc.parameters(), lr=0.001)
train_loss_history = []
val_loss_history = []
train_acc_history = []
val_acc_history = []
num_epochs = 5
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
    train_loss = running_loss / len(train_loader)
    train_acc = 100. * correct / total
    # Валидация
    model.eval()
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            val_running_loss += loss.item()
            _, predicted = outputs.max(1)
            val_total += labels.size(0)
            val_correct += predicted.eq(labels).sum().item()
    val_loss = val_running_loss / len(val_loader)
    val_acc = 100. * val_correct / val_total
    train_loss_history.append(train_loss)
    val_loss_history.append(val_loss)
    train_acc_history.append(train_acc)
    val_acc_history.append(val_acc)
    print(f"Эпоха [{epoch+1}/{num_epochs}] "
          f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}% | "
          f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}%")

```

```python
torch.save(model.state_dict(), "resnet18_mushrooms_frozen.pth")
```

теперь разморозим веса и посмотрим что будет

```python
for param in model.parameters():
    param.requires_grad = True
optimizer_ft = optim.Adam(model.parameters(), lr=0.0001)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer_ft, mode='min', patience=2, factor=0.5)
```

```python

ft_train_loss_history = []
ft_val_loss_history = []
ft_train_acc_history = []
ft_val_acc_history = []

num_ft_epochs = 5

for epoch in range(num_ft_epochs):
    # трэин
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer_ft.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer_ft.step()
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
    ft_train_loss = running_loss / len(train_loader)
    ft_train_acc = 100. * correct / total

    # валидация
    model.eval()
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            val_running_loss += loss.item()
            _, predicted = outputs.max(1)
            val_total += labels.size(0)
            val_correct += predicted.eq(labels).sum().item()
    ft_val_loss = val_running_loss / len(val_loader)
    ft_val_acc = 100. * val_correct / val_total
    ft_train_loss_history.append(ft_train_loss)
    ft_val_loss_history.append(ft_val_loss)
    ft_train_acc_history.append(ft_train_acc)
    ft_val_acc_history.append(ft_val_acc)
    scheduler.step(ft_val_loss)

    print(f"[FT] Epoch [{epoch+1}/{num_ft_epochs}] "
          f"Train Loss: {ft_train_loss:.4f} | Train Acc: {ft_train_acc:.2f}% | "
          f"Val Loss: {ft_val_loss:.4f} | Val Acc: {ft_val_acc:.2f}%")
```

```python
torch.save(model.state_dict(), "resnet18_mushrooms_finetuned.pth")
from google.colab import files
files.download("resnet18_mushrooms_finetuned.pth")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
#График Accuracy
axes[0].plot(range(1, num_epochs + 1), train_acc_history, 'o--', color='blue', alpha=0.6, label='Train (Frozen)')
axes[0].plot(range(1, num_epochs + 1), val_acc_history, 's--', color='red', alpha=0.6, label='Val (Frozen)')
axes[0].plot(range(num_epochs + 1, num_epochs + num_ft_epochs + 1), ft_train_acc_history, 'o-', color='blue', label='Train (Fine-tuned)')
axes[0].plot(range(num_epochs + 1, num_epochs + num_ft_epochs + 1), ft_val_acc_history, 's-', color='red', label='Val (Fine-tuned)')
axes[0].axvline(x=num_epochs + 0.5, color='green', linestyle=':', label='Разморозка')
axes[0].set_title('Accuracy: Frozen vs Fine-tuned')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy (%)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

#График Loss
axes[1].plot(range(1, num_epochs + 1), train_loss_history, 'o--', color='blue', alpha=0.6, label='Train (Frozen)')
axes[1].plot(range(1, num_epochs + 1), val_loss_history, 's--', color='red', alpha=0.6, label='Val (Frozen)')
axes[1].plot(range(num_epochs + 1, num_epochs + num_ft_epochs + 1), ft_train_loss_history, 'o-', color='blue', label='Train (Fine-tuned)')
axes[1].plot(range(num_epochs + 1, num_epochs + num_ft_epochs + 1), ft_val_loss_history, 's-', color='red', label='Val (Fine-tuned)')
axes[1].axvline(x=num_epochs + 0.5, color='green', linestyle=':', label='Разморозка')
axes[1].set_title('Loss: Frozen vs Fine-tuned')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle('Transfer Learning: 2-Stage Training History', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
